```python
import networkx as nx
import matplotlib.pyplot as plt
```

```python
# new graph with no edges or nodes
G = nx.Graph()
```

```python
# add our 10 nodes from krackhardt kite 
G.add_nodes_from(range(1, 10))
```

```python
# add the 18 edges from krackhardt kite 
#The traditional labeling is: Andre=1, Beverley=2, Carol=3, Diane=4, Ed=5, Fernando=6, Garth=7, Heather=8, Ike=9, Jane=10.
kite_edges = [
    (1, 2), (1, 3), (1, 4), (1, 6),
    (2, 4), (2, 5), (2, 7),
    (3, 4), (3, 6),
    (4, 5), (4, 6), (4, 7),
    (5, 7),
    (6, 7), (6, 8),
    (7, 8),
    (8, 9),
    (9, 10)]
```

```python
# add our edges into our graph
G.add_edges_from(kite_edges)
```

```python
# draw the graph
nx.draw(G, with_labels=True, node_color='lightblue')
plt.show()
```
